```python
import matplotlib.pyplot as plt
import numpy as np

# Generate data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Draw a chart
plt.figure(figsize=(6,4))
plt.plot(x, y, label="sin(x)", color="blue")
plt.title("A Simple Sine Curve")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True)
plt.show()
```
